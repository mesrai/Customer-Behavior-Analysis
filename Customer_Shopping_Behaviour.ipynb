```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Setup Connection (Using your details from the image)
username = "root"
password = "root" # Ensure this matches your Workbench password
host = "localhost"
port = "3306"
database = "customer_behaviour"

engine = create_engine(f"mysql+mysqlconnector://{username}:{password}@{host}:{port}/{database}")

# 2. Load the Raw Data
df = pd.read_csv("customer_shopping_behavior.csv")

# 3. DATA CLEANING SECTION (Apply your changes here)
# Example cleaning steps to make the data SQL-friendly:
df_cleaned = df.copy()

# Remove spaces from column names (Best practice for SQL)
df_cleaned.columns = [c.replace(' ', '_').lower() for c in df_cleaned.columns]

# Handle any missing values or data type fixes
df_cleaned = df_cleaned.dropna()

# 4. Upload the CLEANED DataFrame to MySQL
# We use a new table name 'cleaned_shopping_data' to distinguish it
try:
    df_cleaned.to_sql(name='cleaned_shopping_data', 
                      con=engine, 
                      if_exists='replace', 
                      index=False)
    print("Success! The CLEANED data is now available in MySQL Workbench.")
except Exception as e:
    print(f"Connection Error: {e}")
```
